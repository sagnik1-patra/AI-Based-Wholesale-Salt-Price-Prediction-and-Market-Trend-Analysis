```python

import json
import warnings
from pathlib import Path

import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
import yaml

from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)
from tensorflow.keras import Sequential
from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout
)
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ====================================================
# 1. PROJECT CONFIGURATION
# ====================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Wholesale Salt Price Prediction and Market Trend Analysis"
)

DATA_PATH = BASE_DIR / (
    "weekly_wholesale_price_salt_loose_upto-2012.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

SEED = 42
LOOKBACK = 6
FUTURE_WEEKS = 12

AIS_POPULATION = 8
AIS_GENERATIONS = 6
AIS_CLONES = 3
AIS_MUTATION_RATE = 0.25

NN_EPOCHS = 35
NN_BATCH_SIZE = 32

rng = np.random.default_rng(SEED)

np.random.seed(SEED)
tf.random.set_seed(SEED)

sns.set_theme(style="whitegrid")

# ====================================================
# 2. HELPER FUNCTIONS
# ====================================================

def save_plot(filename):
    plt.tight_layout()
    plt.savefig(
        BASE_DIR / filename,
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()


def evaluate(y_true, y_pred):
    return {
        "MAE": float(
            mean_absolute_error(y_true, y_pred)
        ),
        "MSE": float(
            mean_squared_error(y_true, y_pred)
        ),
        "RMSE": float(
            np.sqrt(
                mean_squared_error(y_true, y_pred)
            )
        ),
        "R2": float(
            r2_score(y_true, y_pred)
        )
    }


def json_safe(value):
    if isinstance(value, dict):
        return {
            str(k): json_safe(v)
            for k, v in value.items()
        }

    if isinstance(value, (list, tuple)):
        return [json_safe(v) for v in value]

    if isinstance(value, np.integer):
        return int(value)

    if isinstance(value, np.floating):
        return float(value)

    if isinstance(value, pd.Timestamp):
        return str(value)

    return value


# ====================================================
# 3. LOAD DATASET
# ====================================================

print("\nLoading Wholesale Salt Price Dataset...")

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATA_PATH}"
    )

df = pd.read_csv(DATA_PATH)
df.columns = df.columns.str.strip()

required = ["Date", "zone", "centre", "salt_l"]

missing_cols = [
    c for c in required if c not in df.columns
]

if missing_cols:
    raise ValueError(
        f"Missing columns: {missing_cols}"
    )

original_rows = len(df)

print("Original shape:", df.shape)
print(df.head())

# ====================================================
# 4. DATA CLEANING
# ====================================================

print("\nCleaning Data...")

df["Date"] = pd.to_datetime(
    df["Date"],
    format="%d-%m-%y",
    errors="coerce"
)

df["salt_l"] = pd.to_numeric(
    df["salt_l"],
    errors="coerce"
)

df["zone"] = (
    df["zone"]
    .astype("string")
    .str.strip()
    .str.upper()
)

df["centre"] = (
    df["centre"]
    .astype("string")
    .str.strip()
    .str.upper()
)

missing_price_rows = int(
    df["salt_l"].isna().sum()
)

df = df.dropna(
    subset=["Date", "zone", "centre"]
).copy()

df.loc[df["salt_l"] <= 0, "salt_l"] = np.nan

# Aggregate duplicate observations.
df = (
    df.groupby(
        ["Date", "zone", "centre"],
        as_index=False
    )["salt_l"]
    .mean()
)

df = df.dropna(subset=["salt_l"])

df = df.sort_values(
    ["zone", "centre", "Date"]
).reset_index(drop=True)

if df.empty:
    raise ValueError("No valid price data.")

print("Cleaned rows:", len(df))

# ====================================================
# 5. FEATURE ENGINEERING
# ====================================================

print("\nGenerating Historical Features...")

df["year"] = df["Date"].dt.year
df["month"] = df["Date"].dt.month

df["week"] = (
    df["Date"]
    .dt.isocalendar()
    .week
    .astype(int)
)

df["quarter"] = df["Date"].dt.quarter

groups = df.groupby(
    ["zone", "centre"],
    sort=False
)

lag_cols = []

for lag in range(1, LOOKBACK + 1):
    name = f"lag_{lag}"

    df[name] = groups["salt_l"].shift(lag)
    lag_cols.append(name)

df["days_since_previous"] = (
    groups["Date"]
    .diff()
    .dt.days
)

model_df = df.dropna(
    subset=lag_cols + ["days_since_previous"]
).copy()

model_df = model_df.sort_values(
    ["Date", "zone", "centre"]
).reset_index(drop=True)

if len(model_df) < 50:
    raise ValueError(
        "Not enough observations for model training."
    )

numeric_features = [
    "year",
    "month",
    "week",
    "quarter",
    "days_since_previous"
] + lag_cols

categorical_features = [
    "zone",
    "centre"
]

feature_cols = (
    numeric_features + categorical_features
)

# ====================================================
# 6. CHRONOLOGICAL DATA SPLIT
# ====================================================

unique_dates = sorted(
    model_df["Date"].unique()
)

if len(unique_dates) < 15:
    raise ValueError(
        "Not enough distinct dates for validation."
    )

# Final 20% of dates are held out for testing.
test_idx = int(len(unique_dates) * 0.80)
test_idx = min(
    max(test_idx, 1),
    len(unique_dates) - 1
)

test_start = pd.Timestamp(
    unique_dates[test_idx]
)

train_all = model_df[
    model_df["Date"] < test_start
].copy()

test_df = model_df[
    model_df["Date"] >= test_start
].copy()

# Use last 20% of training-period dates
# for AIS hyperparameter validation.
train_dates = sorted(
    train_all["Date"].unique()
)

val_idx = int(len(train_dates) * 0.80)
val_idx = min(
    max(val_idx, 1),
    len(train_dates) - 1
)

val_start = pd.Timestamp(
    train_dates[val_idx]
)

train_df = train_all[
    train_all["Date"] < val_start
].copy()

val_df = train_all[
    train_all["Date"] >= val_start
].copy()

if any(
    x.empty for x in [
        train_df, val_df, test_df
    ]
):
    raise ValueError("One of the data splits is empty.")

X_train = train_df[feature_cols]
y_train = train_df["salt_l"].to_numpy()

X_val = val_df[feature_cols]
y_val = val_df["salt_l"].to_numpy()

X_full = train_all[feature_cols]
y_full = train_all["salt_l"].to_numpy()

X_test = test_df[feature_cols]
y_test = test_df["salt_l"].to_numpy()

print("\nTrain:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))

# ====================================================
# 7. PREPROCESSING PIPELINE
# ====================================================

def create_preprocessor():
    return ColumnTransformer([
        (
            "numeric",
            StandardScaler(),
            numeric_features
        ),
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        )
    ])


def build_rf(params):
    return Pipeline([
        (
            "preprocessor",
            create_preprocessor()
        ),
        (
            "model",
            RandomForestRegressor(
                n_estimators=int(
                    params["n_estimators"]
                ),
                max_depth=int(
                    params["max_depth"]
                ),
                min_samples_split=int(
                    params["min_samples_split"]
                ),
                min_samples_leaf=int(
                    params["min_samples_leaf"]
                ),
                max_features=float(
                    params["max_features"]
                ),
                random_state=SEED,
                n_jobs=-1
            )
        )
    ])


# ====================================================
# 8. ARTIFICIAL IMMUNE SYSTEM (AIS)
# ====================================================

print("\nStarting AIS Optimization...")

# Each antibody represents one set of
# Random Forest hyperparameters.
#
# Affinity is based on validation RMSE.
# Lower RMSE means higher affinity.

PARAM_BOUNDS = {
    "n_estimators": (50, 180),
    "max_depth": (4, 24),
    "min_samples_split": (2, 12),
    "min_samples_leaf": (1, 6),
    "max_features": (0.4, 1.0)
}

INTEGER_PARAMS = {
    "n_estimators",
    "max_depth",
    "min_samples_split",
    "min_samples_leaf"
}


def random_antibody():
    antibody = {}

    for name, (low, high) in PARAM_BOUNDS.items():

        if name in INTEGER_PARAMS:
            antibody[name] = int(
                rng.integers(low, high + 1)
            )
        else:
            antibody[name] = float(
                rng.uniform(low, high)
            )

    return antibody


def mutate_antibody(antibody, strength):
    mutated = antibody.copy()

    for name, (low, high) in PARAM_BOUNDS.items():

        if rng.random() < AIS_MUTATION_RATE:

            span = high - low

            delta = rng.normal(
                0,
                span * strength
            )

            value = np.clip(
                mutated[name] + delta,
                low,
                high
            )

            if name in INTEGER_PARAMS:
                value = int(round(value))
            else:
                value = float(value)

            mutated[name] = value

    return mutated


def antibody_key(antibody):
    return tuple(
        round(float(antibody[name]), 5)
        for name in PARAM_BOUNDS
    )


fitness_cache = {}


def calculate_affinity(antibody):
    key = antibody_key(antibody)

    if key in fitness_cache:
        return fitness_cache[key]

    model = build_rf(antibody)

    model.fit(X_train, y_train)

    predictions = model.predict(X_val)

    rmse = float(
        np.sqrt(
            mean_squared_error(
                y_val,
                predictions
            )
        )
    )

    fitness_cache[key] = rmse

    return rmse


population = [
    random_antibody()
    for _ in range(AIS_POPULATION)
]

best_antibody = None
best_rmse = float("inf")

optimization_history = []

for generation in range(
    1, AIS_GENERATIONS + 1
):

    print(
        f"\nAIS Generation "
        f"{generation}/{AIS_GENERATIONS}"
    )

    evaluated = []

    for antibody in population:
        rmse = calculate_affinity(antibody)

        evaluated.append(
            (rmse, antibody.copy())
        )

    evaluated.sort(key=lambda x: x[0])

    # Keep best antibodies.
    elite_count = max(
        2, AIS_POPULATION // 3
    )

    elites = evaluated[:elite_count]

    # Clone and hypermutate elite antibodies.
    clones = []

    for rank, (rmse, antibody) in enumerate(elites):

        # Higher-ranked antibodies receive
        # smaller mutations.
        strength = 0.08 + 0.12 * (
            rank / max(1, elite_count - 1)
        )

        for _ in range(AIS_CLONES):

            clone = mutate_antibody(
                antibody,
                strength
            )

            clone_rmse = calculate_affinity(
                clone
            )

            clones.append(
                (clone_rmse, clone)
            )

    combined = evaluated + clones
    combined.sort(key=lambda x: x[0])

    generation_best_rmse = combined[0][0]
    generation_best_params = combined[0][1]

    if generation_best_rmse < best_rmse:
        best_rmse = generation_best_rmse
        best_antibody = (
            generation_best_params.copy()
        )

    optimization_history.append({
        "Generation": generation,
        "Best_RMSE": best_rmse,
        "Generation_Best_RMSE": (
            generation_best_rmse
        ),
        "Population_Mean_RMSE": float(
            np.mean([
                item[0] for item in evaluated
            ])
        )
    })

    print("Best RMSE:", round(best_rmse, 5))
    print("Best Parameters:", best_antibody)

    # Immune memory: retain the best.
    next_population = [
        item[1].copy()
        for item in combined[:elite_count]
    ]

    # Introduce diversity with random antibodies.
    while len(next_population) < AIS_POPULATION:

        if rng.random() < 0.35:
            next_population.append(
                random_antibody()
            )
        else:
            parent = combined[
                int(rng.integers(
                    0,
                    min(5, len(combined))
                ))
            ][1]

            next_population.append(
                mutate_antibody(
                    parent,
                    strength=0.20
                )
            )

    population = next_population


ais_history_df = pd.DataFrame(
    optimization_history
)

ais_history_df.to_csv(
    BASE_DIR / "ais_optimization_history.csv",
    index=False
)

print("\nAIS Optimization Completed.")
print("Best Antibody:", best_antibody)

# ====================================================
# 9. AIS OPTIMIZATION GRAPH
# ====================================================

plt.figure(figsize=(11, 5))

plt.plot(
    ais_history_df["Generation"],
    ais_history_df["Best_RMSE"],
    marker="o",
    label="Best Validation RMSE"
)

plt.plot(
    ais_history_df["Generation"],
    ais_history_df["Population_Mean_RMSE"],
    marker="s",
    linestyle="--",
    label="Population Mean RMSE"
)

plt.title("AIS Clonal Selection Optimization")
plt.xlabel("Generation")
plt.ylabel("Validation RMSE")
plt.legend()

save_plot("ais_optimization_graph.png")

# ====================================================
# 10. TRAIN AIS-OPTIMIZED RANDOM FOREST
# ====================================================

print("\nTraining AIS-Optimized Random Forest...")

ais_rf = build_rf(best_antibody)

ais_rf.fit(X_full, y_full)

rf_pred = np.maximum(
    ais_rf.predict(X_test),
    0
)

joblib.dump(
    {
        "model": ais_rf,
        "features": feature_cols,
        "best_parameters": best_antibody,
        "lookback": LOOKBACK,
        "algorithm": "AIS-Clonal Selection"
    },
    BASE_DIR / "ais_salt_price_model.pkl"
)

# ====================================================
# 11. PREPARE AIS-OPTIMIZED NEURAL NETWORK
# ====================================================

print("\nPreparing Neural Network...")

# The AIS search above optimizes Random Forest.
# We use its selected max_depth parameter
# to configure neural network width.
#
# This is a heuristic transfer of AIS-selected
# settings, not independent NN optimization.

hidden_units = int(
    np.clip(
        best_antibody["max_depth"] * 4,
        16,
        128
    )
)

nn_preprocessor = create_preprocessor()

X_nn_full = nn_preprocessor.fit_transform(
    X_full
)

X_nn_test = nn_preprocessor.transform(
    X_test
)

# Dense neural networks require dense arrays.
if hasattr(X_nn_full, "toarray"):
    X_nn_full = X_nn_full.toarray()

if hasattr(X_nn_test, "toarray"):
    X_nn_test = X_nn_test.toarray()

X_nn_full = np.asarray(
    X_nn_full,
    dtype=np.float32
)

X_nn_test = np.asarray(
    X_nn_test,
    dtype=np.float32
)

target_scaler = StandardScaler()

y_nn_full = target_scaler.fit_transform(
    y_full.reshape(-1, 1)
).astype(np.float32)

# ====================================================
# 12. TRAIN NEURAL NETWORK
# ====================================================

print("\nTraining AIS-Configured Neural Network...")

nn_model = Sequential([
    Input(shape=(X_nn_full.shape[1],)),
    Dense(hidden_units, activation="relu"),
    Dropout(0.20),
    Dense(
        max(8, hidden_units // 2),
        activation="relu"
    ),
    Dense(1)
])

nn_model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

# Chronological validation by row order:
# X_full was created from date-sorted records.
history = nn_model.fit(
    X_nn_full,
    y_nn_full,
    epochs=NN_EPOCHS,
    batch_size=NN_BATCH_SIZE,
    validation_split=0.15,
    shuffle=False,
    callbacks=[
        EarlyStopping(
            monitor="val_loss",
            patience=6,
            restore_best_weights=True
        )
    ],
    verbose=1
)

nn_model.save(
    str(
        BASE_DIR / "ais_salt_price_model.h5"
    )
)

joblib.dump(
    {
        "preprocessor": nn_preprocessor,
        "target_scaler": target_scaler,
        "features": feature_cols
    },
    BASE_DIR / "ais_nn_scaler.pkl"
)

nn_scaled_pred = nn_model.predict(
    X_nn_test,
    verbose=0
)

nn_pred = np.maximum(
    target_scaler.inverse_transform(
        nn_scaled_pred
    ).flatten(),
    0
)

# ====================================================
# 13. MODEL EVALUATION
# ====================================================

rf_metrics = evaluate(
    y_test,
    rf_pred
)

nn_metrics = evaluate(
    y_test,
    nn_pred
)

comparison = pd.DataFrame([
    {
        "Model": "AIS Random Forest",
        **rf_metrics
    },
    {
        "Model": "AIS Configured NN",
        **nn_metrics
    }
])

comparison.to_csv(
    BASE_DIR / "ais_model_comparison.csv",
    index=False
)

print("\nAIS MODEL COMPARISON")
print(comparison.to_string(index=False))

best_model_name = (
    "AIS Random Forest"
    if rf_metrics["RMSE"] <= nn_metrics["RMSE"]
    else "AIS Configured NN"
)

best_pred = (
    rf_pred
    if best_model_name == "AIS Random Forest"
    else nn_pred
)

print("\nBest Test Model:", best_model_name)

# ====================================================
# 14. AIS ACCURACY GRAPH
# ====================================================

# R2 is used instead of classification accuracy.

fig, axes = plt.subplots(
    1, 2, figsize=(13, 5)
)

sns.barplot(
    data=comparison,
    x="Model",
    y="R2",
    ax=axes[0],
    color="steelblue"
)

axes[0].set_title("AIS Model R2 Scores")
axes[0].set_ylabel("R2 Score")
axes[0].tick_params(
    axis="x",
    labelrotation=15
)

sns.barplot(
    data=comparison,
    x="Model",
    y="RMSE",
    ax=axes[1],
    color="orange"
)

axes[1].set_title("AIS Model RMSE")
axes[1].set_ylabel("RMSE")
axes[1].tick_params(
    axis="x",
    labelrotation=15
)

save_plot("ais_accuracy_graph.png")

# ====================================================
# 15. AIS HEATMAP
# ====================================================

corr_cols = (
    numeric_features + ["salt_l"]
)

corr = train_all[corr_cols].corr()

plt.figure(figsize=(12, 9))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title(
    "AIS Salt Price Feature Correlation Heatmap"
)

save_plot("ais_heatmap.png")

# ====================================================
# 16. AIS COMPARISON GRAPH
# ====================================================

n_show = min(150, len(y_test))

plt.figure(figsize=(14, 6))

plt.plot(
    y_test[:n_show],
    label="Actual Price",
    linewidth=2
)

plt.plot(
    rf_pred[:n_show],
    label="AIS Random Forest",
    linestyle="--"
)

plt.plot(
    nn_pred[:n_show],
    label="AIS Configured NN",
    linestyle=":"
)

plt.title("AIS Model Prediction Comparison")
plt.xlabel("Test Observation")
plt.ylabel("Wholesale Salt Price")
plt.legend()

save_plot("ais_comparison_graph.png")

# ====================================================
# 17. AIS RESULT CSV
# ====================================================

result_df = test_df[
    ["Date", "zone", "centre", "salt_l"]
].copy()

result_df = result_df.rename(
    columns={
        "salt_l": "Actual_Price"
    }
)

result_df["AIS_RF_Prediction"] = rf_pred
result_df["AIS_NN_Prediction"] = nn_pred

result_df["Best_Model_Prediction"] = (
    best_pred
)

result_df["Absolute_Error"] = np.abs(
    result_df["Actual_Price"]
    - result_df["Best_Model_Prediction"]
)

result_df["Squared_Error"] = (
    result_df["Actual_Price"]
    - result_df["Best_Model_Prediction"]
) ** 2

result_df.to_csv(
    BASE_DIR / "ais_result.csv",
    index=False
)

print("\nSaved AIS Result CSV")
print(result_df.head())

# ====================================================
# 18. AIS RESULT GRAPH
# ====================================================

plt.figure(figsize=(10, 7))

plt.scatter(
    result_df["Actual_Price"],
    result_df["Best_Model_Prediction"],
    alpha=0.5,
    label="AIS Predictions"
)

minimum = min(
    result_df["Actual_Price"].min(),
    result_df["Best_Model_Prediction"].min()
)

maximum = max(
    result_df["Actual_Price"].max(),
    result_df["Best_Model_Prediction"].max()
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    "r--",
    label="Perfect Prediction"
)

plt.title(
    f"AIS Actual vs Predicted - {best_model_name}"
)

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.legend()

save_plot("ais_result_graph.png")

# ====================================================
# 19. AIS FUTURE FORECAST
# ====================================================

print("\nGenerating AIS Future Forecasts...")

prediction_rows = []

for (zone, centre), market in df.groupby(
    ["zone", "centre"],
    sort=False
):

    market = market.sort_values("Date")

    if len(market) < LOOKBACK:
        continue

    prices = market["salt_l"].tolist()
    last_date = market["Date"].iloc[-1]

    for step in range(1, FUTURE_WEEKS + 1):

        future_date = last_date + pd.Timedelta(
            weeks=1
        )

        feature_row = {
            "year": future_date.year,
            "month": future_date.month,
            "week": int(
                future_date.isocalendar().week
            ),
            "quarter": future_date.quarter,
            "days_since_previous": 7,
            "zone": zone,
            "centre": centre
        }

        for lag in range(1, LOOKBACK + 1):
            feature_row[f"lag_{lag}"] = (
                prices[-lag]
            )

        future_input = pd.DataFrame(
            [feature_row]
        )[feature_cols]

        if best_model_name == "AIS Random Forest":

            forecast = float(
                ais_rf.predict(
                    future_input
                )[0]
            )

        else:

            nn_input = nn_preprocessor.transform(
                future_input
            )

            if hasattr(nn_input, "toarray"):
                nn_input = nn_input.toarray()

            nn_input = np.asarray(
                nn_input,
                dtype=np.float32
            )

            scaled_prediction = (
                nn_model.predict(
                    nn_input,
                    verbose=0
                )
            )

            forecast = float(
                target_scaler.inverse_transform(
                    scaled_prediction
                )[0, 0]
            )

        forecast = max(0.0, forecast)

        prediction_rows.append({
            "zone": zone,
            "centre": centre,
            "Forecast_Date": future_date,
            "Forecast_Week": step,
            "AIS_Predicted_Price": forecast,
            "Model": best_model_name
        })

        prices.append(forecast)
        last_date = future_date


prediction_df = pd.DataFrame(
    prediction_rows
)

if prediction_df.empty:
    raise ValueError(
        "No eligible markets for forecasting."
    )

prediction_df.to_csv(
    BASE_DIR / "ais_prediction.csv",
    index=False
)

print(prediction_df.head(15))

# ====================================================
# 20. AIS PREDICTION GRAPH
# ====================================================

market_counts = (
    df.groupby(["zone", "centre"])
    .size()
    .sort_values(ascending=False)
)

selected_zone, selected_centre = (
    market_counts.index[0]
)

historical_market = df[
    (df["zone"] == selected_zone)
    & (df["centre"] == selected_centre)
].sort_values("Date").tail(30)

future_market = prediction_df[
    (prediction_df["zone"] == selected_zone)
    & (prediction_df["centre"] == selected_centre)
].sort_values("Forecast_Date")

plt.figure(figsize=(14, 6))

plt.plot(
    historical_market["Date"],
    historical_market["salt_l"],
    marker="o",
    label="Historical Price"
)

plt.plot(
    future_market["Forecast_Date"],
    future_market["AIS_Predicted_Price"],
    marker="o",
    linestyle="--",
    label="AIS Forecast"
)

plt.axvline(
    historical_market["Date"].max(),
    color="gray",
    linestyle=":",
    label="Forecast Start"
)

plt.title(
    f"AIS 12-Week Salt Price Forecast - "
    f"{selected_centre}"
)

plt.xlabel("Date")
plt.ylabel("Wholesale Salt Price")
plt.xticks(rotation=45)
plt.legend()

save_plot("ais_prediction_graph.png")

# ====================================================
# 21. AIS TRAINING LOSS GRAPH
# ====================================================

plt.figure(figsize=(10, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title(
    "AIS-Configured Neural Network Training Loss"
)

plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.legend()

save_plot("ais_training_loss_graph.png")

# ====================================================
# 22. SAVE AIS CONFIG YAML
# ====================================================

config = {
    "project": (
        "AI-Based Wholesale Salt Price Prediction "
        "and Market Trend Analysis"
    ),
    "algorithm": "Artificial Immune System",
    "ais_variant": "Clonal Selection",
    "dataset": str(DATA_PATH),
    "target": "salt_l",
    "lookback": LOOKBACK,
    "future_weeks": FUTURE_WEEKS,
    "population": AIS_POPULATION,
    "generations": AIS_GENERATIONS,
    "clones_per_elite": AIS_CLONES,
    "mutation_rate": AIS_MUTATION_RATE,
    "best_parameters": best_antibody,
    "validation_rmse": best_rmse,
    "best_test_model": best_model_name,
    "features": feature_cols,
    "output_directory": str(BASE_DIR)
}

with open(
    BASE_DIR / "ais_config.yaml",
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        json_safe(config),
        file,
        sort_keys=False
    )

# ====================================================
# 23. SAVE AIS METADATA JSON
# ====================================================

metadata = {
    "project_name": (
        "AI-Based Wholesale Salt Price Prediction "
        "and Market Trend Analysis"
    ),
    "algorithm": "AIS Clonal Selection",
    "original_rows": original_rows,
    "missing_price_rows": missing_price_rows,
    "valid_rows": int(len(df)),
    "model_ready_rows": int(len(model_df)),
    "train_rows": int(len(train_df)),
    "validation_rows": int(len(val_df)),
    "test_rows": int(len(test_df)),
    "test_start_date": str(test_start.date()),
    "ais_settings": {
        "population": AIS_POPULATION,
        "generations": AIS_GENERATIONS,
        "clones_per_elite": AIS_CLONES,
        "mutation_rate": AIS_MUTATION_RATE
    },
    "best_parameters": best_antibody,
    "best_validation_rmse": best_rmse,
    "random_forest_metrics": rf_metrics,
    "neural_network_metrics": nn_metrics,
    "best_test_model": best_model_name,
    "forecast_weeks": FUTURE_WEEKS,
    "forecast_note": (
        "Forecasts begin after the last historical "
        "observation for each market."
    )
}

with open(
    BASE_DIR / "ais_model_metadata.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_safe(metadata),
        file,
        indent=4
    )

# ====================================================
# 24. VERIFY OUTPUT FILES
# ====================================================

output_files = [
    "ais_salt_price_model.pkl",
    "ais_salt_price_model.h5",
    "ais_nn_scaler.pkl",
    "ais_config.yaml",
    "ais_model_metadata.json",
    "ais_accuracy_graph.png",
    "ais_heatmap.png",
    "ais_comparison_graph.png",
    "ais_result.csv",
    "ais_result_graph.png",
    "ais_prediction.csv",
    "ais_prediction_graph.png",
    "ais_optimization_graph.png",
    "ais_training_loss_graph.png",
    "ais_model_comparison.csv",
    "ais_optimization_history.csv"
]

print("\n" + "=" * 65)
print("AIS SALT PRICE PREDICTION PROJECT COMPLETED")
print("=" * 65)

print("\nAIS Best Parameters:")
print(json.dumps(
    json_safe(best_antibody),
    indent=4
))

print("\nAIS Random Forest Metrics:")
print(json.dumps(rf_metrics, indent=4))

print("\nAIS Neural Network Metrics:")
print(json.dumps(nn_metrics, indent=4))

print("\nBest Test Model:", best_model_name)

print("\nGenerated Files:")

for filename in output_files:

    file_path = BASE_DIR / filename

    if file_path.exists():
        print("[SAVED]", filename)
    else:
        print("[MISSING]", filename)

print("\nOutput Directory:")
print(BASE_DIR)

print("\nAIS Analysis Finished Successfully!")
```
