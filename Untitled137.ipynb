```python

import os
import json
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml
import tensorflow as tf

from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input, LSTM, Dense, Dropout
)
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ==================================================
# 1. PATH CONFIGURATION
# ==================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Wholesale Salt Price Prediction and Market Trend Analysis"
)

DATA_PATH = BASE_DIR / (
    "weekly_wholesale_price_salt_loose_upto-2012.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

SEED = 42
LOOKBACK = 6
FUTURE_WEEKS = 12

np.random.seed(SEED)
tf.random.set_seed(SEED)

sns.set_theme(style="whitegrid")

# ==================================================
# 2. LOAD DATASET
# ==================================================

print("\nLoading Salt Price Dataset...")

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATA_PATH}"
    )

df = pd.read_csv(DATA_PATH)

df.columns = df.columns.str.strip()

print("\nDataset Shape:", df.shape)
print("\nDataset Columns:", df.columns.tolist())
print("\nFirst 5 Records:")
print(df.head())

required = ["Date", "zone", "centre", "salt_l"]

for col in required:
    if col not in df.columns:
        raise ValueError(f"Missing column: {col}")

original_rows = len(df)
missing_prices = int(
    pd.to_numeric(
        df["salt_l"], errors="coerce"
    ).isna().sum()
)

# ==================================================
# 3. DATA CLEANING
# ==================================================

print("\nCleaning Dataset...")

df["Date"] = pd.to_datetime(
    df["Date"],
    format="%d-%m-%y",
    errors="coerce"
)

df["salt_l"] = pd.to_numeric(
    df["salt_l"],
    errors="coerce"
)

df["zone"] = (
    df["zone"]
    .astype("string")
    .str.strip()
    .str.upper()
)

df["centre"] = (
    df["centre"]
    .astype("string")
    .str.strip()
    .str.upper()
)

df = df.dropna(
    subset=["Date", "zone", "centre"]
)

df.loc[df["salt_l"] <= 0, "salt_l"] = np.nan

# Combine duplicate records.
df = (
    df.groupby(
        ["Date", "zone", "centre"],
        as_index=False
    )["salt_l"]
    .mean()
)

# Do not invent missing target prices.
df = df.dropna(subset=["salt_l"])

df = df.sort_values(
    ["zone", "centre", "Date"]
).reset_index(drop=True)

print("Cleaned Dataset:", df.shape)

if df.empty:
    raise ValueError("No valid price records.")

# ==================================================
# 4. FEATURE ENGINEERING
# ==================================================

print("\nGenerating Features...")

df["year"] = df["Date"].dt.year
df["month"] = df["Date"].dt.month
df["week"] = (
    df["Date"].dt.isocalendar().week.astype(int)
)
df["quarter"] = df["Date"].dt.quarter

groups = df.groupby(
    ["zone", "centre"],
    sort=False
)

lag_cols = []

for lag in range(1, LOOKBACK + 1):
    col = f"lag_{lag}"
    df[col] = groups["salt_l"].shift(lag)
    lag_cols.append(col)

df["days_since_previous"] = (
    groups["Date"].diff().dt.days
)

model_df = df.dropna(
    subset=lag_cols + ["days_since_previous"]
).copy()

model_df = model_df.sort_values(
    ["Date", "zone", "centre"]
).reset_index(drop=True)

if len(model_df) < 30:
    raise ValueError(
        "Insufficient rows after lag generation."
    )

numeric_features = [
    "year",
    "month",
    "week",
    "quarter",
    "days_since_previous"
] + lag_cols

categorical_features = ["zone", "centre"]

feature_cols = (
    numeric_features + categorical_features
)

# ==================================================
# 5. CHRONOLOGICAL TRAIN / TEST SPLIT
# ==================================================

unique_dates = sorted(
    model_df["Date"].unique()
)

split_index = int(len(unique_dates) * 0.8)

split_index = max(
    1,
    min(split_index, len(unique_dates) - 1)
)

cutoff = pd.Timestamp(
    unique_dates[split_index]
)

train_df = model_df[
    model_df["Date"] < cutoff
].copy()

test_df = model_df[
    model_df["Date"] >= cutoff
].copy()

if train_df.empty or test_df.empty:
    raise ValueError("Invalid train/test split.")

X_train = train_df[feature_cols]
X_test = test_df[feature_cols]

y_train = train_df["salt_l"].to_numpy()
y_test = test_df["salt_l"].to_numpy()

print("\nTraining Samples:", len(train_df))
print("Testing Samples:", len(test_df))
print("Testing Starts:", cutoff.date())

# ==================================================
# 6. RANDOM FOREST MODEL
# ==================================================

print("\nTraining Random Forest...")

preprocessor = ColumnTransformer([
    (
        "numeric",
        StandardScaler(),
        numeric_features
    ),
    (
        "categorical",
        OneHotEncoder(
            handle_unknown="ignore"
        ),
        categorical_features
    )
])

rf_model = Pipeline([
    ("preprocessor", preprocessor),
    (
        "model",
        RandomForestRegressor(
            n_estimators=150,
            max_depth=16,
            min_samples_leaf=2,
            random_state=SEED,
            n_jobs=-1
        )
    )
])

rf_model.fit(X_train, y_train)

rf_pred = np.maximum(
    rf_model.predict(X_test),
    0
)

joblib.dump(
    {
        "model": rf_model,
        "features": feature_cols,
        "lookback": LOOKBACK
    },
    BASE_DIR / "salt_price_model.pkl"
)

# ==================================================
# 7. LSTM DATA PREPARATION
# ==================================================

print("\nPreparing LSTM Data...")

price_scaler = StandardScaler()

price_scaler.fit(
    train_df[["salt_l"]]
)

def make_sequences(data, scaler):
    arr = data[
        [f"lag_{i}" for i in range(LOOKBACK, 0, -1)]
    ].to_numpy(dtype=np.float32)

    scaled = scaler.transform(
        arr.reshape(-1, 1)
    )

    return scaled.reshape(
        len(arr), LOOKBACK, 1
    ).astype(np.float32)


X_lstm_train = make_sequences(
    train_df, price_scaler
)

X_lstm_test = make_sequences(
    test_df, price_scaler
)

y_lstm_train = price_scaler.transform(
    y_train.reshape(-1, 1)
)

# ==================================================
# 8. BUILD AND TRAIN LSTM
# ==================================================

print("\nTraining LSTM...")

lstm_model = Sequential([
    Input(shape=(LOOKBACK, 1)),
    LSTM(64, return_sequences=True),
    Dropout(0.2),
    LSTM(32),
    Dropout(0.2),
    Dense(16, activation="relu"),
    Dense(1)
])

lstm_model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

history = lstm_model.fit(
    X_lstm_train,
    y_lstm_train,
    epochs=35,
    batch_size=32,
    validation_split=0.15,
    shuffle=False,
    callbacks=[
        EarlyStopping(
            monitor="val_loss",
            patience=6,
            restore_best_weights=True
        )
    ],
    verbose=1
)

lstm_model.save(
    str(BASE_DIR / "salt_price_model.h5")
)

joblib.dump(
    price_scaler,
    BASE_DIR / "lstm_scaler.pkl"
)

lstm_scaled_pred = lstm_model.predict(
    X_lstm_test,
    verbose=0
)

lstm_pred = np.maximum(
    price_scaler.inverse_transform(
        lstm_scaled_pred
    ).flatten(),
    0
)

# ==================================================
# 9. MODEL EVALUATION
# ==================================================

def evaluate(actual, predicted):
    return {
        "MAE": float(
            mean_absolute_error(actual, predicted)
        ),
        "MSE": float(
            mean_squared_error(actual, predicted)
        ),
        "RMSE": float(
            np.sqrt(
                mean_squared_error(actual, predicted)
            )
        ),
        "R2": float(
            r2_score(actual, predicted)
        )
    }


rf_metrics = evaluate(y_test, rf_pred)
lstm_metrics = evaluate(y_test, lstm_pred)

comparison = pd.DataFrame([
    {"Model": "Random Forest", **rf_metrics},
    {"Model": "LSTM", **lstm_metrics}
])

comparison.to_csv(
    BASE_DIR / "model_comparison.csv",
    index=False
)

print("\nMODEL PERFORMANCE")
print(comparison.to_string(index=False))

best_model_name = (
    "Random Forest"
    if rf_metrics["RMSE"] <= lstm_metrics["RMSE"]
    else "LSTM"
)

best_pred = (
    rf_pred
    if best_model_name == "Random Forest"
    else lstm_pred
)

print("\nBest Model:", best_model_name)

# ==================================================
# 10. ACCURACY GRAPH
# ==================================================

print("\nGenerating Accuracy Graph...")

# Regression does not have classification accuracy.
# R2 and RMSE are used instead.

fig, axes = plt.subplots(
    1, 2, figsize=(13, 5)
)

sns.barplot(
    data=comparison,
    x="Model",
    y="R2",
    ax=axes[0],
    color="steelblue"
)

axes[0].set_title(
    "Model Performance - R2 Score"
)
axes[0].set_ylabel("R2 Score")
axes[0].axhline(0, color="black", linewidth=0.8)

sns.barplot(
    data=comparison,
    x="Model",
    y="RMSE",
    ax=axes[1],
    color="orange"
)

axes[1].set_title(
    "Model Prediction Error - RMSE"
)
axes[1].set_ylabel("RMSE (Lower is Better)")

plt.tight_layout()

plt.savefig(
    BASE_DIR / "accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==================================================
# 11. HEATMAP
# ==================================================

print("\nGenerating Correlation Heatmap...")

heatmap_columns = (
    numeric_features + ["salt_l"]
)

corr = train_df[
    heatmap_columns
].corr()

plt.figure(figsize=(12, 9))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title(
    "Wholesale Salt Price Feature Correlation"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==================================================
# 12. COMPARISON GRAPH
# ==================================================

print("\nGenerating Comparison Graph...")

# Compare the first 150 test observations.
n_show = min(150, len(y_test))

plt.figure(figsize=(14, 6))

plt.plot(
    y_test[:n_show],
    label="Actual Price",
    linewidth=2
)

plt.plot(
    rf_pred[:n_show],
    label="Random Forest Prediction",
    linestyle="--"
)

plt.plot(
    lstm_pred[:n_show],
    label="LSTM Prediction",
    linestyle=":"
)

plt.title(
    "Actual vs Predicted Wholesale Salt Prices"
)

plt.xlabel("Test Observation")
plt.ylabel("Wholesale Salt Price")
plt.legend()
plt.tight_layout()

plt.savefig(
    BASE_DIR / "comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==================================================
# 13. RESULT CSV
# ==================================================

print("\nSaving Result CSV...")

result_df = test_df[
    ["Date", "zone", "centre", "salt_l"]
].copy()

result_df = result_df.rename(
    columns={"salt_l": "Actual_Price"}
)

result_df["RF_Predicted_Price"] = rf_pred
result_df["LSTM_Predicted_Price"] = lstm_pred
result_df["Best_Model_Prediction"] = best_pred

result_df["Absolute_Error"] = np.abs(
    result_df["Actual_Price"]
    - result_df["Best_Model_Prediction"]
)

result_df["Squared_Error"] = (
    result_df["Actual_Price"]
    - result_df["Best_Model_Prediction"]
) ** 2

result_df.to_csv(
    BASE_DIR / "result.csv",
    index=False
)

print(result_df.head())

# ==================================================
# 14. RESULT GRAPH
# ==================================================

print("\nGenerating Result Graph...")

plt.figure(figsize=(14, 6))

plt.scatter(
    result_df["Actual_Price"],
    result_df["Best_Model_Prediction"],
    alpha=0.5,
    label="Predictions"
)

minimum = min(
    result_df["Actual_Price"].min(),
    result_df["Best_Model_Prediction"].min()
)

maximum = max(
    result_df["Actual_Price"].max(),
    result_df["Best_Model_Prediction"].max()
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    "r--",
    label="Perfect Prediction"
)

plt.title(
    f"Prediction Results - {best_model_name}"
)

plt.xlabel("Actual Wholesale Price")
plt.ylabel("Predicted Wholesale Price")
plt.legend()
plt.tight_layout()

plt.savefig(
    BASE_DIR / "result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==================================================
# 15. FUTURE PRICE PREDICTION
# ==================================================

print("\nGenerating Future Predictions...")

# Generate recursive 12-week forecasts for
# each market using the selected best model.
#
# Forecast dates are weekly after the final
# available observation for each market.
#
# This is an experimental historical forecast,
# not a prediction of current 2026 prices.

prediction_rows = []

for (zone, centre), market in df.groupby(
    ["zone", "centre"],
    sort=False
):
    market = market.sort_values("Date")

    if len(market) < LOOKBACK:
        continue

    prices = market["salt_l"].tolist()
    last_date = market["Date"].iloc[-1]

    for step in range(1, FUTURE_WEEKS + 1):

        future_date = last_date + pd.Timedelta(
            weeks=1
        )

        # Most recent six observed/predicted prices.
        previous_prices = prices[-LOOKBACK:]

        if best_model_name == "Random Forest":

            previous_gap = 7

            feature_row = {
                "year": future_date.year,
                "month": future_date.month,
                "week": int(
                    future_date.isocalendar().week
                ),
                "quarter": future_date.quarter,
                "days_since_previous": previous_gap,
                "zone": zone,
                "centre": centre
            }

            for lag in range(1, LOOKBACK + 1):
                feature_row[f"lag_{lag}"] = (
                    previous_prices[-lag]
                )

            future_input = pd.DataFrame(
                [feature_row]
            )[feature_cols]

            forecast_price = float(
                rf_model.predict(future_input)[0]
            )

        else:

            sequence = np.asarray(
                previous_prices,
                dtype=np.float32
            ).reshape(-1, 1)

            scaled_sequence = (
                price_scaler.transform(sequence)
                .reshape(1, LOOKBACK, 1)
            )

            scaled_forecast = (
                lstm_model.predict(
                    scaled_sequence,
                    verbose=0
                )
            )

            forecast_price = float(
                price_scaler.inverse_transform(
                    scaled_forecast
                )[0, 0]
            )

        forecast_price = max(
            0.0, forecast_price
        )

        prediction_rows.append({
            "zone": zone,
            "centre": centre,
            "Forecast_Date": future_date,
            "Forecast_Week": step,
            "Predicted_Price": forecast_price,
            "Model": best_model_name
        })

        prices.append(forecast_price)
        last_date = future_date

prediction_df = pd.DataFrame(
    prediction_rows
)

if prediction_df.empty:
    raise ValueError(
        "No markets have enough historical data for forecasting."
    )

prediction_df.to_csv(
    BASE_DIR / "prediction.csv",
    index=False
)

print("\nFuture Predictions:")
print(prediction_df.head(15))

# ==================================================
# 16. PREDICTION GRAPH
# ==================================================

print("\nGenerating Prediction Graph...")

# Plot one representative market, avoiding
# mixing different market price scales.

market_counts = (
    df.groupby(["zone", "centre"])
    .size()
    .sort_values(ascending=False)
)

selected_zone, selected_centre = (
    market_counts.index[0]
)

historical_market = df[
    (df["zone"] == selected_zone)
    & (df["centre"] == selected_centre)
].sort_values("Date").tail(30)

future_market = prediction_df[
    (prediction_df["zone"] == selected_zone)
    & (prediction_df["centre"] == selected_centre)
].sort_values("Forecast_Date")

plt.figure(figsize=(14, 6))

plt.plot(
    historical_market["Date"],
    historical_market["salt_l"],
    marker="o",
    label="Historical Price"
)

plt.plot(
    future_market["Forecast_Date"],
    future_market["Predicted_Price"],
    marker="o",
    linestyle="--",
    label="Forecast Price"
)

plt.axvline(
    historical_market["Date"].max(),
    color="gray",
    linestyle=":",
    label="Forecast Start"
)

plt.title(
    f"12-Week Salt Price Forecast - "
    f"{selected_centre} ({selected_zone})"
)

plt.xlabel("Date")
plt.ylabel("Wholesale Salt Price")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()

plt.savefig(
    BASE_DIR / "prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==================================================
# 17. SAVE YAML CONFIGURATION
# ==================================================

config = {
    "project_name": (
        "AI-Based Wholesale Salt Price Prediction "
        "and Market Trend Analysis"
    ),
    "dataset": str(DATA_PATH),
    "target": "salt_l",
    "random_seed": SEED,
    "lookback": LOOKBACK,
    "future_weeks": FUTURE_WEEKS,
    "test_split": "Chronological 80/20 by date",
    "features": feature_cols,
    "models": [
        "Random Forest",
        "LSTM"
    ],
    "best_model": best_model_name,
    "output_directory": str(BASE_DIR)
}

with open(
    BASE_DIR / "config.yaml",
    "w",
    encoding="utf-8"
) as file:
    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

# ==================================================
# 18. SAVE JSON METADATA
# ==================================================

metadata = {
    "project": (
        "AI-Based Wholesale Salt Price Prediction "
        "and Market Trend Analysis"
    ),
    "original_records": original_rows,
    "missing_price_records": missing_prices,
    "valid_records": int(len(df)),
    "model_ready_records": int(len(model_df)),
    "training_records": int(len(train_df)),
    "testing_records": int(len(test_df)),
    "date_range": {
        "start": str(df["Date"].min().date()),
        "end": str(df["Date"].max().date())
    },
    "random_forest_metrics": rf_metrics,
    "lstm_metrics": lstm_metrics,
    "best_model": best_model_name,
    "forecast_weeks": FUTURE_WEEKS,
    "forecast_note": (
        "Recursive forecasts from each market's "
        "last available historical observation. "
        "Not current-market price estimates."
    )
}

with open(
    BASE_DIR / "model_metadata.json",
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        metadata,
        file,
        indent=4
    )

# ==================================================
# 19. SAVE TRAINING HISTORY GRAPH
# ==================================================

plt.figure(figsize=(10, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("LSTM Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.legend()
plt.tight_layout()

plt.savefig(
    BASE_DIR / "training_loss_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==================================================
# 20. FINAL OUTPUT
# ==================================================

print("\n" + "=" * 60)
print("SALT PRICE ML PROJECT COMPLETED")
print("=" * 60)

print("\nBest Performing Model:", best_model_name)

print("\nRandom Forest Metrics:")
print(json.dumps(rf_metrics, indent=4))

print("\nLSTM Metrics:")
print(json.dumps(lstm_metrics, indent=4))

print("\nGenerated Files:")

output_files = [
    "salt_price_model.pkl",
    "salt_price_model.h5",
    "lstm_scaler.pkl",
    "config.yaml",
    "model_metadata.json",
    "model_comparison.csv",
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "training_loss_graph.png"
]

for filename in output_files:
    path = BASE_DIR / filename
    if path.exists():
        print(f"[SAVED] {filename}")
    else:
        print(f"[MISSING] {filename}")

print("\nAll available results saved in:")
print(BASE_DIR)
```
