```python

from pathlib import Path
import json
import warnings

import joblib
import numpy as np
import pandas as pd
import yaml
import tensorflow as tf

from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ==========================================
# 1. PROJECT PATHS
# ==========================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Wholesale Salt Price Prediction and Market Trend Analysis"
)

CSV_PATH = BASE_DIR / "weekly_wholesale_price_salt_loose_upto-2012.csv"

BASE_DIR.mkdir(parents=True, exist_ok=True)

PKL_PATH = BASE_DIR / "salt_price_model.pkl"
H5_PATH = BASE_DIR / "salt_price_model.h5"
YAML_PATH = BASE_DIR / "config.yaml"
JSON_PATH = BASE_DIR / "model_metadata.json"
COMPARISON_PATH = BASE_DIR / "model_comparison.csv"

SEED = 42
LOOKBACK = 6
TEST_RATIO = 0.20
EPOCHS = 35
BATCH_SIZE = 32

np.random.seed(SEED)
tf.random.set_seed(SEED)

# ==========================================
# 2. LOAD DATASET
# ==========================================

print("\nLoading dataset...")

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {CSV_PATH}"
    )

df = pd.read_csv(CSV_PATH)

print("Original shape:", df.shape)
print("Columns:", df.columns.tolist())
print(df.head())

required_columns = [
    "Date", "zone", "centre", "salt_l"
]

missing_columns = [
    c for c in required_columns
    if c not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

original_rows = len(df)
original_missing = int(df["salt_l"].isna().sum())

# ==========================================
# 3. DATA CLEANING
# ==========================================

print("\nCleaning dataset...")

# Dataset dates use DD-MM-YY format.
# Example: 06-01-93 = 6 January 1993.
df["Date"] = pd.to_datetime(
    df["Date"],
    format="%d-%m-%y",
    errors="coerce"
)

df["salt_l"] = pd.to_numeric(
    df["salt_l"],
    errors="coerce"
)

df["zone"] = df["zone"].astype("string")
df["centre"] = (
    df["centre"]
    .astype("string")
    .str.strip()
    .str.upper()
)

df = df.dropna(
    subset=["Date", "zone", "centre"]
).copy()

# Invalid or nonpositive prices are excluded.
df.loc[df["salt_l"] <= 0, "salt_l"] = np.nan

# Keep one record per market and date.
df = (
    df.groupby(
        ["Date", "zone", "centre"],
        as_index=False
    )["salt_l"]
    .mean()
)

df = df.sort_values(
    ["centre", "zone", "Date"]
).reset_index(drop=True)

# Preserve the historical series without
# inventing prices for missing observations.
df = df.dropna(subset=["salt_l"]).copy()

if df.empty:
    raise ValueError(
        "No valid salt price observations found."
    )

print("Valid observations:", len(df))
print("Missing original prices:", original_missing)

# ==========================================
# 4. FEATURE ENGINEERING
# ==========================================

print("\nGenerating historical features...")

# Each prediction uses only earlier observed
# prices from the same centre and zone.
group_cols = ["zone", "centre"]
groups = df.groupby(group_cols, sort=False)

lag_columns = []

for lag in range(1, LOOKBACK + 1):
    col = f"lag_{lag}"
    df[col] = groups["salt_l"].shift(lag)
    lag_columns.append(col)

df["year"] = df["Date"].dt.year
df["month"] = df["Date"].dt.month
df["week"] = df["Date"].dt.isocalendar().week.astype(int)
df["quarter"] = df["Date"].dt.quarter

df["days_since_previous"] = (
    groups["Date"]
    .diff()
    .dt.days
)

# Rows without sufficient historical prices
# cannot be used for supervised training.
model_df = df.dropna(
    subset=lag_columns + ["days_since_previous"]
).copy()

model_df = model_df.sort_values(
    ["Date", "zone", "centre"]
).reset_index(drop=True)

if len(model_df) < 30:
    raise ValueError(
        "Insufficient observations after creating lag features."
    )

print("Training-ready rows:", len(model_df))

# ==========================================
# 5. CHRONOLOGICAL TRAIN-TEST SPLIT
# ==========================================

unique_dates = np.array(
    sorted(model_df["Date"].unique())
)

if len(unique_dates) < 10:
    raise ValueError(
        "Insufficient distinct dates for time-based testing."
    )

split_index = int(
    len(unique_dates) * (1 - TEST_RATIO)
)

split_index = max(
    1, min(split_index, len(unique_dates) - 1)
)

cutoff_date = pd.Timestamp(
    unique_dates[split_index]
)

train_df = model_df[
    model_df["Date"] < cutoff_date
].copy()

test_df = model_df[
    model_df["Date"] >= cutoff_date
].copy()

if train_df.empty or test_df.empty:
    raise ValueError(
        "Chronological split produced an empty dataset."
    )

print("Training rows:", len(train_df))
print("Testing rows:", len(test_df))
print("Test starts:", cutoff_date.date())

# ==========================================
# 6. RANDOM FOREST PREPROCESSING
# ==========================================

categorical_features = [
    "zone",
    "centre"
]

numerical_features = [
    "year",
    "month",
    "week",
    "quarter",
    "days_since_previous"
] + lag_columns

feature_columns = (
    categorical_features + numerical_features
)

X_train = train_df[feature_columns]
X_test = test_df[feature_columns]

y_train = train_df["salt_l"].to_numpy()
y_test = test_df["salt_l"].to_numpy()

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="most_frequent")
    ),
    (
        "encoder",
        OneHotEncoder(
            handle_unknown="ignore"
        )
    )
])

preprocessor = ColumnTransformer([
    (
        "numerical",
        numeric_pipeline,
        numerical_features
    ),
    (
        "categorical",
        categorical_pipeline,
        categorical_features
    )
])

# ==========================================
# 7. RANDOM FOREST TRAINING
# ==========================================

print("\nTraining Random Forest...")

rf_model = Pipeline([
    ("preprocessor", preprocessor),
    (
        "regressor",
        RandomForestRegressor(
            n_estimators=150,
            max_depth=16,
            min_samples_leaf=2,
            random_state=SEED,
            n_jobs=-1
        )
    )
])

rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_test)

# ==========================================
# 8. SAVE PKL MODEL
# ==========================================

joblib.dump(
    {
        "model": rf_model,
        "features": feature_columns,
        "lookback": LOOKBACK,
        "target": "salt_l",
        "date_format": "%d-%m-%y",
        "test_start_date": str(cutoff_date.date())
    },
    PKL_PATH
)

print("Saved:", PKL_PATH.name)

# ==========================================
# 9. PREPARE LSTM SEQUENCES
# ==========================================

print("\nPreparing LSTM input...")

# LSTM uses previous prices in chronological
# order, from oldest to newest.
# The scaler is fitted only on training data.

price_scaler = StandardScaler()

price_scaler.fit(
    train_df[["salt_l"]]
)

def make_lstm_sequences(data, scaler):
    sequences = []

    for _, row in data.iterrows():
        history = [
            row[f"lag_{lag}"]
            for lag in range(LOOKBACK, 0, -1)
        ]

        sequences.append(history)

    sequences = np.asarray(
        sequences,
        dtype=np.float32
    )

    original_shape = sequences.shape

    scaled = scaler.transform(
        sequences.reshape(-1, 1)
    )

    return scaled.reshape(
        original_shape[0],
        original_shape[1],
        1
    ).astype(np.float32)


X_lstm_train = make_lstm_sequences(
    train_df, price_scaler
)

X_lstm_test = make_lstm_sequences(
    test_df, price_scaler
)

y_lstm_train = price_scaler.transform(
    y_train.reshape(-1, 1)
).astype(np.float32)

# ==========================================
# 10. BUILD LSTM MODEL
# ==========================================

print("\nBuilding LSTM model...")

lstm_model = Sequential([
    Input(shape=(LOOKBACK, 1)),

    LSTM(
        64,
        return_sequences=True
    ),

    Dropout(0.2),

    LSTM(32),

    Dropout(0.2),

    Dense(
        16,
        activation="relu"
    ),

    Dense(1)
])

lstm_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

lstm_model.summary()

# ==========================================
# 11. TRAIN LSTM
# ==========================================

print("\nTraining LSTM...")

callbacks = [
    EarlyStopping(
        monitor="val_loss",
        patience=6,
        restore_best_weights=True
    )
]

history = lstm_model.fit(
    X_lstm_train,
    y_lstm_train,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.15,
    shuffle=False,
    callbacks=callbacks,
    verbose=1
)

# ==========================================
# 12. SAVE H5 MODEL
# ==========================================

lstm_model.save(str(H5_PATH))

# Save the LSTM input scaler separately.
joblib.dump(
    price_scaler,
    BASE_DIR / "lstm_scaler.pkl"
)

print("Saved:", H5_PATH.name)

# ==========================================
# 13. GENERATE LSTM PREDICTIONS
# ==========================================

lstm_scaled_predictions = lstm_model.predict(
    X_lstm_test,
    verbose=0
)

lstm_predictions = (
    price_scaler.inverse_transform(
        lstm_scaled_predictions
    ).flatten()
)

# Salt prices cannot be negative.
rf_predictions = np.maximum(
    rf_predictions, 0
)

lstm_predictions = np.maximum(
    lstm_predictions, 0
)

# ==========================================
# 14. MODEL EVALUATION
# ==========================================

def evaluate_model(y_true, y_pred):
    return {
        "MAE": float(
            mean_absolute_error(y_true, y_pred)
        ),
        "MSE": float(
            mean_squared_error(y_true, y_pred)
        ),
        "RMSE": float(
            np.sqrt(
                mean_squared_error(y_true, y_pred)
            )
        ),
        "R2": float(
            r2_score(y_true, y_pred)
        )
    }


rf_metrics = evaluate_model(
    y_test, rf_predictions
)

lstm_metrics = evaluate_model(
    y_test, lstm_predictions
)

comparison = pd.DataFrame([
    {
        "Model": "Random Forest",
        **rf_metrics
    },
    {
        "Model": "LSTM",
        **lstm_metrics
    }
])

comparison.to_csv(
    COMPARISON_PATH,
    index=False
)

print("\nMODEL COMPARISON")
print(comparison.to_string(index=False))

# ==========================================
# 15. SAVE CONFIG.YAML
# ==========================================

config = {
    "project": {
        "name": (
            "AI-Based Wholesale Salt Price "
            "Prediction and Market Trend Analysis"
        ),
        "task": "Time-Series Regression",
        "target": "salt_l",
        "dataset": str(CSV_PATH)
    },

    "dataset": {
        "original_rows": original_rows,
        "missing_price_rows": original_missing,
        "valid_price_rows": int(len(df)),
        "model_ready_rows": int(len(model_df)),
        "train_rows": int(len(train_df)),
        "test_rows": int(len(test_df)),
        "test_start_date": str(cutoff_date.date())
    },

    "features": {
        "categorical": categorical_features,
        "numerical": numerical_features,
        "lookback": LOOKBACK
    },

    "training": {
        "test_ratio": TEST_RATIO,
        "split_method": "chronological_by_date",
        "random_seed": SEED,
        "epochs": EPOCHS,
        "batch_size": BATCH_SIZE
    },

    "models": {
        "random_forest": {
            "estimators": 150,
            "max_depth": 16,
            "min_samples_leaf": 2
        },
        "lstm": {
            "units": [64, 32],
            "optimizer": "adam",
            "loss": "mse"
        }
    },

    "output_files": {
        "random_forest": PKL_PATH.name,
        "lstm": H5_PATH.name,
        "lstm_scaler": "lstm_scaler.pkl",
        "metadata": JSON_PATH.name,
        "comparison": COMPARISON_PATH.name
    }
}

with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:
    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )

print("Saved:", YAML_PATH.name)

# ==========================================
# 16. SAVE MODEL_METADATA.JSON
# ==========================================

best_model = (
    "Random Forest"
    if rf_metrics["RMSE"] <= lstm_metrics["RMSE"]
    else "LSTM"
)

metadata = {
    "project_name": (
        "AI-Based Wholesale Salt Price "
        "Prediction and Market Trend Analysis"
    ),

    "dataset_information": {
        "original_rows": original_rows,
        "valid_price_rows": int(len(df)),
        "missing_price_rows": original_missing,
        "model_ready_rows": int(len(model_df)),
        "training_rows": int(len(train_df)),
        "testing_rows": int(len(test_df)),
        "total_zones": int(df["zone"].nunique()),
        "total_centres": int(df["centre"].nunique()),
        "minimum_date": str(df["Date"].min().date()),
        "maximum_date": str(df["Date"].max().date())
    },

    "target_column": "salt_l",

    "features": feature_columns,

    "evaluation": {
        "random_forest": rf_metrics,
        "lstm": lstm_metrics
    },

    "best_model_by_rmse": best_model,

    "methodology": {
        "split": "Chronological train/test split",
        "missing_prices": (
            "Excluded missing target observations"
        ),
        "historical_inputs": (
            "Six previous observed prices per market"
        ),
        "lstm_inputs": (
            "Historical prices only; no market identifiers"
        ),
        "forecast_type": (
            "One-step prediction for an observed target date"
        )
    },

    "model_files": {
        "random_forest": PKL_PATH.name,
        "lstm": H5_PATH.name,
        "lstm_scaler": "lstm_scaler.pkl"
    }
}

with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        metadata,
        file,
        indent=4,
        ensure_ascii=False
    )

print("Saved:", JSON_PATH.name)

# ==========================================
# 17. VERIFY SAVED MODELS
# ==========================================

print("\nVerifying saved models...")

loaded_rf = joblib.load(PKL_PATH)["model"]

sample_rf = loaded_rf.predict(
    X_test.iloc[:1]
)

loaded_lstm = tf.keras.models.load_model(
    str(H5_PATH),
    compile=False
)

sample_lstm = loaded_lstm.predict(
    X_lstm_test[:1],
    verbose=0
)

print(
    "Random Forest sample prediction:",
    float(sample_rf[0])
)

print(
    "LSTM sample prediction:",
    float(
        price_scaler.inverse_transform(
            sample_lstm
        )[0, 0]
    )
)

# ==========================================
# 18. FINAL OUTPUT
# ==========================================

print("\n" + "=" * 55)
print("PROJECT TRAINING COMPLETED")
print("=" * 55)

print("\nBest model:", best_model)

print("\nRandom Forest Metrics:")
print(json.dumps(rf_metrics, indent=4))

print("\nLSTM Metrics:")
print(json.dumps(lstm_metrics, indent=4))

print("\nGenerated files:")

for output_file in [
    PKL_PATH,
    H5_PATH,
    BASE_DIR / "lstm_scaler.pkl",
    YAML_PATH,
    JSON_PATH,
    COMPARISON_PATH
]:
    print(output_file)

print("\nAll models and metadata saved successfully.")
```
